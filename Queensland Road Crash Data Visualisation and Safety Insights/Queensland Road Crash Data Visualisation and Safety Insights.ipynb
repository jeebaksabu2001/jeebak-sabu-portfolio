### 3. Cleaning the Data

```python
# Count exact‐row duplicates
dupe_count = crash_data.duplicated().sum()
print(dupe_count)
```

```python
crash_data['Crash_Ref_Number'].nunique()
```

```python
# Count missing (NaN) values in each column
null_counts = crash_data.isna().sum().sort_values(ascending=False)

# Display the result
print("Missing values per column:")
print(null_counts)
```

```python
# Filling the required catgorical columns to Unkwown for better distribution
cols_to_fill = [
    "Crash_Street_Intersecting",
    "State_Road_Name",
    "DCA_Key_Approach_Dir",
    "Crash_Street",
    "Crash_Speed_Limit",
    "Crash_Road_Vert_Align"
]

# For each column in that list, replacing NaN to "Unknown"
for col in cols_to_fill:
    if col in crash_data.columns:
        # Convert to string type first (so that "Unknown" is valid)
        crash_data[col] = crash_data[col].astype(str)
        # Wherever the value is "nan" (the string representation of NaN) or an actual NaN, set it to "Unknown"
        crash_data[col] = crash_data[col].replace("nan", np.nan).fillna("Unknown")

# Verify that those columns no longer have any missing values
print(crash_data[cols_to_fill].isna().sum())
```

```python
# I want to convert Crash_month to numberical(January=1, February=2,...,December=12) so that I can use it for possible analysis
# Striping whitespace and making sure everything is a string
crash_data["Crash_Month"] = crash_data["Crash_Month"].astype(str).str.strip()

# Build a month‐name to integer map 
month_map_full = {m: i for i, m in enumerate([
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
], start=1)}
month_map_abbrev = {m: i for i, m in enumerate([
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
], start=1)}

# Parsing each entry into an integer month:
def parse_month(val):
    if pd.isna(val) or val.lower() in ["", "nan", "unknown", "n/a"]:
        return np.nan
    v = val.strip()
    # If it already looks numeric then coerce
    if v.isdigit():
        num = int(v)
        return num if 1 <= num <= 12 else np.nan
    # If it matches a full month name then return that
    for name, idx in month_map_full.items():
        if v.lower() == name.lower():
            return idx
    # If it matches a three‐letter abbreviation
    for abbr, idx in month_map_abbrev.items():
        if v.lower() == abbr.lower():
            return idx
    # Otherwise cannot parse
    return np.nan
```

```python
crash_data["Crash_Month_Int"] = crash_data["Crash_Month"].apply(parse_month).astype("Int64")
```

```python
# 6. Verify how many months failed to parse (eg: became <NA>)
num_na_months = crash_data["Crash_Month_Int"].isna().sum()
print(f"\nRows with unparsed Crash_Month: {num_na_months} out of {len(crash_data)}")
```

```python
# Overwriting Crash_Month with the integer version:
crash_data["Crash_Month"] = crash_data["Crash_Month_Int"]

# 8. Droping the helper column
crash_data.drop(columns=["Crash_Month_Int"], inplace=True)
```
