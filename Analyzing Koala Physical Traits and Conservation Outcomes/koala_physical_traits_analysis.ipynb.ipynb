```python
import numpy as np

# Function to replace outliers with the median within gender groups
def replace_outliers_with_median_by_gender(column):
    Q1 = column.quantile(0.25)
    Q3 = column.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    median = column.median()

    # Replace outliers with the median
    column = np.where((column < lower_bound) | (column > upper_bound), median, column)
    
    return column

# Apply the function to all numeric columns to replace outliers only based on gender
numeric_cols = koala_data.select_dtypes(include=[np.number])
for col in numeric_cols.columns:
    koala_data[col] = koala_data.groupby('gender')[col].transform(replace_outliers_with_median_by_gender)

# Check the result by printing summary statistics
print(koala_data.describe())

# Plot histograms to check the effect of replacing outliers
koala_data.hist(figsize=(12, 10), bins=20, edgecolor='black', grid=False)
plt.tight_layout()
plt.show()
```

### Is the mean head length of the Koalas significantly different from 92.0 mm?

```python
from scipy import stats

# Extract the 'head_length' data (in centimeters) and remove any missing values
head_length_data = koala_data['head_length'].dropna()

# Convert the value of 92 mm to 9.2 cm
comparison_value = 9.2

# Perform a one-sample t-test
t_stat, p_value = stats.ttest_1samp(head_length_data, comparison_value)

# Calculate mean of head_length
mean_head_length = head_length_data.mean()

# Output the t-statistic, p-value, and mean head length
print(f"T-Statistic: {t_stat}")
print(f"P-Value: {p_value}")
print(f"Mean Head Length: {mean_head_length}")

# Conclusion based on p-value
if p_value < 0.05:
    print("Reject the null hypothesis: The mean head length is significantly different from 9.2 cm (92.0 mm).")
else:
    print("Fail to reject the null hypothesis: The mean head length is not significantly different from 9.2 cm (92.0 mm).")
```

### Do male and female Koalas have significantly different mean head lengths?

```python
from scipy import stats

# Separate the head length data for males and females
male_head_length = koala_data[koala_data['gender'] == 'male']['head_length'].dropna()
female_head_length = koala_data[koala_data['gender'] == 'female']['head_length'].dropna()

# Perform an independent two-sample t-test
t_stat, p_value = stats.ttest_ind(male_head_length, female_head_length, equal_var=False)  # Set equal_var=False if variances are not equal

# Output the t-statistic and p-value
print(f"T-Statistic: {t_stat}")
print(f"P-Value: {p_value}")

# Calculate the mean head length for each gender
mean_male_head_length = male_head_length.mean()
mean_female_head_length = female_head_length.mean()

print(f"Mean Head Length of Males: {mean_male_head_length}")
print(f"Mean Head Length of Females: {mean_female_head_length}")

# Conclusion based on p-value
if p_value < 0.05:
    print("Reject the null hypothesis: Male and female koalas have significantly different mean head lengths.")
else:
    print("Fail to reject the null hypothesis: Male and female koalas do not have significantly different mean head lengths.")
```

### Do environmental factors such as state affect Koala’s physical characteristics?

```python
from scipy import stats

# Select the traits and environmental factor (state/region)
koala_anova_data = koala_data[['region', 'head_length', 'foot_length', 'Paw Size']].dropna()

# Perform ANOVA for each physical trait

# ANOVA for Head Length
anova_head_length = stats.f_oneway(
    *[koala_anova_data[koala_anova_data['region'] == region]['head_length'] for region in koala_anova_data['region'].unique()]
)

# ANOVA for Foot Length
anova_foot_length = stats.f_oneway(
    *[koala_anova_data[koala_anova_data['region'] == region]['foot_length'] for region in koala_anova_data['region'].unique()]
)

# ANOVA for Paw Size
anova_paw_size = stats.f_oneway(
    *[koala_anova_data[koala_anova_data['region'] == region]['Paw Size'] for region in koala_anova_data['region'].unique()]
)

# Output the ANOVA results
print("ANOVA Results for Head Length:")
print(f"F-statistic: {anova_head_length.statistic}, P-value: {anova_head_length.pvalue}\n")

print("ANOVA Results for Foot Length:")
print(f"F-statistic: {anova_foot_length.statistic}, P-value: {anova_foot_length.pvalue}\n")

print("ANOVA Results for Paw Size:")
print(f"F-statistic: {anova_paw_size.statistic}, P-value: {anova_paw_size.pvalue}\n")
```
