```python
# Calculating the IQR and identifying outliers
Q1 = data[['value_sales', 'value_cost', 'value_quantity']].quantile(0.25)
Q3 = data[['value_sales', 'value_cost', 'value_quantity']].quantile(0.75)
IQR = Q3 - Q1

# Identifying outliers
outliers = ((data[['value_sales', 'value_cost', 'value_quantity']] < (Q1 - 1.5 * IQR)) |
            (data[['value_sales', 'value_cost', 'value_quantity']] > (Q3 + 1.5 * IQR)))

outliers.sum()  # Summing to see the count of outliers in each column
```

```python
# Splitting positive and negative values to understand the distribution
positive_values = data[['value_sales', 'value_cost', 'value_quantity']] > 0
negative_values = data[['value_sales', 'value_cost', 'value_quantity']] < 0

# Counting positive and negative entries in each column
positive_count = positive_values.sum()
negative_count = negative_values.sum()

positive_count, negative_count
```

```python
# Calculating IQR for positive values only
Q1_positive = data[data['value_sales'] > 0][['value_sales', 'value_cost', 'value_quantity']].quantile(0.25)
Q3_positive = data[data['value_sales'] > 0][['value_sales', 'value_cost', 'value_quantity']].quantile(0.75)
IQR_positive = Q3_positive - Q1_positive

# Capping positive values that exceed the upper threshold
data.loc[data['value_sales'] > (Q3_positive['value_sales'] + 1.5 * IQR_positive['value_sales']), 'value_sales'] = Q3_positive['value_sales'] + 1.5 * IQR_positive['value_sales']
data.loc[data['value_cost'] > (Q3_positive['value_cost'] + 1.5 * IQR_positive['value_cost']), 'value_cost'] = Q3_positive['value_cost'] + 1.5 * IQR_positive['value_cost']
data.loc[data['value_quantity'] > (Q3_positive['value_quantity'] + 1.5 * IQR_positive['value_quantity']), 'value_quantity'] = Q3_positive['value_quantity'] + 1.5 * IQR_positive['value_quantity']
```

```python
# Calculating IQR for negative values only
Q1_negative = data[data['value_sales'] < 0][['value_sales', 'value_cost', 'value_quantity']].quantile(0.25)
Q3_negative = data[data['value_sales'] < 0][['value_sales', 'value_cost', 'value_quantity']].quantile(0.75)
IQR_negative = Q3_negative - Q1_negative
# ged_df['value_quantity'] < (Q1_negative['value_quantity'] - 1.5 * IQR_negative['value_quantity']), 'value_quantity'] = Q1_negative['value_quantity'] - 1.5 * IQR_negative['value_quantity']

# Capping negative values that are below the lower threshold
data.loc[data['value_sales'] < (Q1_negative['value_sales'] - 1.5 * IQR_negative['value_sales']), 'value_sales'] = Q1_negative['value_sales'] - 1.5 * IQR_negative['value_sales']
data.loc[data['value_cost'] < (Q1_negative['value_cost'] - 1.5 * IQR_negative['value_cost']), 'value_cost'] = Q1_negative['value_cost'] - 1.5 * IQR_negative['value_cost']
data.loc[data['value_quantity'] < (Q1_negative['value_quantity'] - 1.5 * IQR_negative['value_quantity']), 'value_quantity'] = Q1_negative['value_quantity'] - 1.5 * IQR_negative['value_quantity']
```

# Section 3 -  Test Sub Sample Differences 

## Question 1
## **Is there a significant difference in the company’s profit margins between the fiscal years 2012 and 2013?**


```python
from scipy.stats import ttest_ind

# Calculate profit margin
data['profit_margin'] = data['value_sales'] - data['value_cost']

# Filter data for the two fiscal years, 2012 and 2013
profit_margin_2012 = data[data['fiscal_year'] == 2012]['profit_margin']
profit_margin_2013 = data[data['fiscal_year'] == 2013]['profit_margin']

# Perform the two-sample t-test
t_stat, p_value = ttest_ind(profit_margin_2012.dropna(), profit_margin_2013.dropna())

# Output the results
print(f"T-test results:\nT-statistic: {t_stat}, P-value: {p_value}")

# Interpret the result
if p_value < 0.05:
    print("There is a statistically significant difference in profit margins between the fiscal years 2012 and 2013.")
else:
    print("There is no statistically significant difference in profit margins between the fiscal years 2012 and 2013.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Calculate the mean and standard deviation of profit margins for 2012 and 2013
mean_2012 = data[data['fiscal_year'] == 2012]['profit_margin'].mean()
mean_2013 = data[data['fiscal_year'] == 2013]['profit_margin'].mean()
n_2012 = data[data['fiscal_year'] == 2012]['profit_margin'].count()
n_2013 = data[data['fiscal_year'] == 2013]['profit_margin'].count()
std_2012 = data[data['fiscal_year'] == 2012]['profit_margin'].std()
std_2013 = data[data['fiscal_year'] == 2013]['profit_margin'].std()

# Calculate 95% confidence intervals for each year
ci_2012 = 1.96 * (std_2012 / np.sqrt(n_2012))
ci_2013 = 1.96 * (std_2013 / np.sqrt(n_2013))
```

```python
# Setting colours
sns.set_theme(style="whitegrid")
palette = ['#a1dab4', '#41b6c4']  

# Create the bar plot with swapped colors and confidence intervals
plt.figure(figsize=(10, 8))
bars = sns.barplot(x=['2012', '2013'], y=[mean_2012, mean_2013], yerr=[ci_2012, ci_2013], capsize=0.2, palette=palette[::-1], edgecolor='black')

# Add a border around the bars for emphasis
for bar in bars.patches:
    bar.set_edgecolor('black')
    bar.set_linewidth(1.5)

# Customize the title, ticks and axis labels with bright, bold fonts
plt.title('Average Profit Margin per Transaction for 2012 and 2013', fontsize=18, weight='bold', color='#333333')
plt.xlabel('Fiscal Year', fontsize=14, weight='bold', color='#333333')
plt.ylabel('Average Profit Margin', fontsize=14, weight='bold', color='#333333')
plt.xticks(fontsize=12, weight='bold', color='#333333')
plt.yticks(fontsize=12, color='#333333')

# Add a grid and set the color for the gridlines
plt.grid(visible=True, which='major', color='#EEEEEE', linewidth=0.8)

# Make the plot borders more pronounced
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['bottom'].set_linewidth(1.2)
plt.gca().spines['left'].set_linewidth(1.2)
plt.gca().set_facecolor('#FAFAFA')

# Show the plot
plt.show()
```

## Question 2
## **Do Returning Customers Have Higher Order Quantities Than New Customers?**

```python
#Calculate the number of orders per customer
customer_order_counts = data['customer_code'].value_counts()

# Determine the average order frequency
average_order_frequency = customer_order_counts.mean()
print(f"Average order frequency per customer: {average_order_frequency}")

# Classify customers based on the average order frequency
# Customers with orders > average are "returning", <= average are "new"
returning_customers = customer_order_counts[customer_order_counts > average_order_frequency].index
new_customers = customer_order_counts[customer_order_counts <= average_order_frequency].index

# Filter data for new and returning customers based on customer_code
new_customer_orders = data[data['customer_code'].isin(new_customers)]['value_quantity']
returning_customer_orders = data[data['customer_code'].isin(returning_customers)]['value_quantity']

# Perform the two-sample t-test
t_stat, p_value = ttest_ind(new_customer_orders.dropna(), returning_customer_orders.dropna())

# Dsiplaying the results
print(f"T-test results:\nT-statistic: {t_stat}, P-value: {p_value}")

# Interpret the result
if p_value < 0.05:
    print("There is a statistically significant difference in customer order quantities between new and returning customers.")
else:
    print("There is no statistically significant difference in customer order quantities between new and returning customers.")
```

## TIME SERIES FORECASTING using SARIMA

```python
data = data_outliers_handled
```

```python
import warnings
warnings.filterwarnings("ignore")

# Aggregate sales data by month
monthly_sales = data.resample('M', on='accounting_date')['value_sales'].sum().reset_index()

# Display the monthly sales data
print(monthly_sales)
```

```python
# Plotting for monthly sales
plt.figure(figsize=(10, 6))
plt.plot(monthly_sales['accounting_date'], monthly_sales['value_sales'], marker='o')
plt.title("Monthly Sales Data")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.show()
```

```python
from statsmodels.tsa.arima.model import ARIMA

# Fit the ARIMA model
model = ARIMA(monthly_sales['value_sales'], order=(1, 1, 1))
model_fit = model.fit()

# Print model summary for details
print(model_fit.summary())
```

```python
import itertools
from statsmodels.tsa.statespace.sarimax import SARIMAX

# Define the p, d, q and seasonal P, D, Q ranges
p = d = q = range(0, 2)
seasonal_p = seasonal_d = seasonal_q = range(0, 2)
seasonal_s = 12

# Generate all possible combinations of (p, d, q) and (P, D, Q, s) combinations
param_combinations = list(itertools.product(p, d, q))
seasonal_param_combinations = list(itertools.product(seasonal_p, seasonal_d, seasonal_q, [seasonal_s]))

# Store results
best_aic = float("inf")
best_params = None
best_seasonal_params = None

# Grid search
for params in param_combinations:
    for seasonal_params in seasonal_param_combinations:
        try:
            # Fit SARIMA model with given parameters
            model = SARIMAX(monthly_sales['value_sales'], order=params, seasonal_order=seasonal_params)
            model_fit = model.fit(disp=False)
            aic = model_fit.aic
            
            # Check if this is the best model so far
            if aic < best_aic:
                best_aic = aic
                best_params = params
                best_seasonal_params = seasonal_params
        except:
            continue

print(f"Best SARIMA Parameters: {best_params}x{best_seasonal_params} with AIC: {best_aic}")
```

```python
# Fit the SARIMA model with the best parameters found from the grid search
optimized_sarima_model = SARIMAX(monthly_sales['value_sales'], 
                                 order=(0, 1, 0), 
                                 seasonal_order=(0, 1, 1, 12))
optimized_model_fit = optimized_sarima_model.fit(disp=False)

# Print summary for the optimized model
print(optimized_model_fit.summary())

# Forecast for the next 12 months (2014)
optimized_forecast = optimized_model_fit.forecast(steps=12)
print("Optimized SARIMA Forecast for 2014:", optimized_forecast)
```
